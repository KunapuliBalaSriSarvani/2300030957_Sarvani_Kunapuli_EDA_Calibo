```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
df = pd.read_csv("messy_ecommerce_15000_student_practice.csv")
print("Dataset Dimensions:", df.shape)
print(df.columns) #To see the columns
print(df.head()) #To see the first 5 rows of the dataset
print(df.dtypes.value_counts()) #To see the data types
print("\nColumns and Data Types:")
print(df.dtypes)
print(df.info()) #To see the summary of the dataset
```

```python
# MISSING VALUES VISUALIZATION
# ============================================================

import matplotlib.pyplot as plt

missing = df.isnull().sum()

# Keep only columns having missing values
missing = missing[missing > 0]
plt.figure(figsize=(12, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    cmap="YlOrRd"
)

plt.title("Missing Values Heatmap", fontweight="bold")
plt.xlabel("Columns")
plt.ylabel("Records")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))
 
sns.heatmap(df.isnull(), cbar=False)
 
plt.title("Missing Values in E-Commerce Dataset")
plt.show()
```

```python
# ============================================================
# MISSING VALUES VISUALIZATION
# ============================================================

import matplotlib.pyplot as plt

missing = df.isnull().sum()

# Keep only columns having missing values
missing = missing[missing > 0]

plt.figure(figsize=(10, 5))

plt.bar(
    missing.index,
    missing.values,
    color="mediumorchid",
    edgecolor="black"
)

plt.title("Missing Values by Column", fontweight="bold")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")

plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

# Show values on top of bars
for i, value in enumerate(missing.values):
    plt.text(
        i,
        value + 5,
        str(value),
        ha="center",
        fontweight="bold"
    )

plt.tight_layout()
plt.show()
```
