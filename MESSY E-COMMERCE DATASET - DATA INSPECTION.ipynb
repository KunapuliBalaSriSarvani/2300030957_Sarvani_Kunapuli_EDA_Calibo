```python
# ============================================================
#        MESSY E-COMMERCE DATASET - DATA INSPECTION
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. LOAD THE DATASET
# ------------------------------------------------------------

df = pd.read_csv("messy_ecommerce_15000_student_practice.csv")

print("Dataset loaded successfully!")


# ============================================================
# 2. DATASET DIMENSIONS
# ============================================================

print("\n" + "="*50)
print("DATASET DIMENSIONS")
print("="*50)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Dataset shape:", df.shape)


# ============================================================
# 3. FIRST FEW RECORDS
# ============================================================

print("\n" + "="*50)
print("FIRST 5 RECORDS")
print("="*50)

display(df.head())


# ============================================================
# 4. DATA TYPES
# ============================================================

print("\n" + "="*50)
print("DATA TYPES")
print("="*50)

print(df.dtypes)


# ============================================================
# 5. MISSING-VALUE ANALYSIS
# ============================================================

print("\n" + "="*50)
print("MISSING-VALUE ANALYSIS")
print("="*50)

missing_values = df.isnull().sum()

missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

missing_analysis = pd.DataFrame({
    "Missing Count": missing_values,
    "Missing Percentage": missing_percentage
})

display(missing_analysis)


# ============================================================
# 6. DUPLICATE-RECORD ANALYSIS
# ============================================================

print("\n" + "="*50)
print("DUPLICATE-RECORD ANALYSIS")
print("="*50)

duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)

if duplicate_count > 0:
    print("\nDuplicate records:")
    display(df[df.duplicated()])


# ============================================================
# 7. STATISTICAL SUMMARY
# ============================================================

print("\n" + "="*50)
print("STATISTICAL SUMMARY")
print("="*50)

display(df.describe())


# ============================================================
# 8. IDENTIFY NUMERICAL COLUMNS
# ============================================================

print("\n" + "="*50)
print("NUMERICAL COLUMNS")
print("="*50)

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print(numerical_columns)


# ============================================================
# 9. IDENTIFY CATEGORICAL COLUMNS
# ============================================================

print("\n" + "="*50)
print("CATEGORICAL COLUMNS")
print("="*50)

categorical_columns = df.select_dtypes(
    include="object"
).columns.tolist()

print(categorical_columns)


# ============================================================
# 10. IDENTIFY DATE COLUMNS
# ============================================================

print("\n" + "="*50)
print("DATE COLUMNS")
print("="*50)

# Convert Order_Date into datetime
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
)

date_columns = df.select_dtypes(
    include="datetime"
).columns.tolist()

print(date_columns)


# ============================================================
# 11. IDENTIFY POSSIBLE OUTLIERS
#     Using IQR METHOD
# ============================================================

print("\n" + "="*50)
print("POSSIBLE OUTLIERS")
print("="*50)

outlier_summary = []

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outlier_summary.append({
        "Column": column,
        "Lower Limit": round(lower_limit, 2),
        "Upper Limit": round(upper_limit, 2),
        "Outlier Count": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df)


# ============================================================
# 12. IDENTIFY INCONSISTENT CATEGORICAL VALUES
# ============================================================

print("\n" + "="*50)
print("CATEGORICAL VALUE ANALYSIS")
print("="*50)

for column in categorical_columns:

    print("\n----------------------------------------")
    print("Column:", column)
    print("----------------------------------------")

    print(
        df[column]
        .value_counts(dropna=False)
    )


# ============================================================
# 13. SPECIFIC CHECK FOR COMMON INCONSISTENCIES
# ============================================================

print("\n" + "="*50)
print("POSSIBLE CATEGORICAL INCONSISTENCIES")
print("="*50)


# Gender
print("\nGender values:")
print(df["Gender"].dropna().unique())


# Payment Method
print("\nPayment Method values:")
print(df["Payment_Method"].dropna().unique())


# Returned
print("\nReturned values:")
print(df["Returned"].dropna().unique())


# ============================================================
# 14. SUMMARY OF DATA QUALITY ISSUES
# ============================================================

print("\n" + "="*50)
print("DATA QUALITY SUMMARY")
print("="*50)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("Duplicate records:", df.duplicated().sum())

print(
    "Total missing values:",
    df.isnull().sum().sum()
)

print(
    "Numerical columns:",
    len(numerical_columns)
)

print(
    "Categorical columns:",
    len(categorical_columns)
)

print(
    "Date columns:",
    len(date_columns)
)
```
